```python
data['Department'] = data['Department'].fillna(data['Department'].mode()[0])
data['Department'].value_counts().sum()
```

```python
data['Residence_Type'] = data['Residence_Type'].fillna(data['Residence_Type'].mode()[0])
```

```python
data['Study_Hours_Per_Day'] = data['Study_Hours_Per_Day'].fillna(data['Study_Hours_Per_Day'].mean())
```

```python
data['Attendance_Percent'] = data['Attendance_Percent'].fillna(data['Attendance_Percent'].mean())
```

```python
data['Previous_GPA'] = data['Previous_GPA'].fillna(data['Previous_GPA'].median())
```

```python
mask = data['Previous_GPA'] <= 5
data.loc[mask,'Backlogs'] = data.loc[mask,'Backlogs'].fillna(data['Backlogs'].mean())
```

```python
mask2 = data['Previous_GPA'] > 5
data.loc[mask2,'Backlogs'] = data.loc[mask2,'Backlogs'].fillna(0)
```
