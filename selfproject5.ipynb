```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
top_paper_gen = df.groupby('Paper Name')['GEN'].max().sort_values(ascending=False).head().index
a=df[df['Paper Name'].isin(top_paper_gen)]
sns.barplot(y="GEN",x="Year",data=a)
plt.subplot(2,2,2)
top_paper_obc = df.groupby('Paper Name')['OBC'].max().sort_values(ascending=False).head().index
b=df[df['Paper Name'].isin(top_paper_obc)]
sns.barplot(y="OBC",x="Year",data=b)
plt.subplot(2,2,3)
top_paper_scst = df.groupby('Paper Name')['SC/ST'].max().sort_values(ascending=False).head().index
c=df[df['Paper Name'].isin(top_paper_scst)]
sns.barplot(y="SC/ST",x="Year",data=c)
```

```python
a1=df[df['Paper Code']=="EE"]
a1
```

```python
plt.plot('Year','GEN',data=a1,marker='o')
plt.title('Year wise UR category cit-off for EE branch')
```

```python
a1 = df[df['Paper Code']=="EE"].sort_values('Year')

plt.figure(figsize=(7,5))
plt.plot(a1['Year'], a1['GEN'], marker='o', label='GEN')
plt.plot(a1['Year'], a1['OBC'], marker='o', label='OBC')
plt.plot(a1['Year'], a1['SC/ST'], marker='o', label='SC/ST')
plt.title('EE Cutoff by Category, 2015-2020')
plt.xlabel('Year')
plt.ylabel('Cutoff Marks')
plt.legend()
plt.grid(alpha=0.3)
```
