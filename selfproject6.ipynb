```python
total_year = df.groupby('Year')[['Applied','Appeared','Not Appeared']].sum().reset_index()
total_year
```

```python
long_df = total_year.melt(id_vars="Year",var_name="Status",value_name="Count")
long_df
```

```python
plt.figure(figsize=(9,8))
sns.barplot(data=long_df,y='Count',x='Year',hue='Status')
plt.tight_layout()
```
